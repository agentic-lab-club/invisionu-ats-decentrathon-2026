```python
audio_filepath =  "ssstik.io_1774782080131"
```

```python

client = Groq(api_key=os.getenv("GROQ_WHISPER_API_KEY"))

def transcribe_audio(filepath: str):
    try:
        with open(filepath, "rb") as file:
            print(f"Отправка файла {filepath} в Groq API...")
            # Вызов метода транскрипции Whisper
            transcription = client.audio.transcriptions.create(
                file=(filepath, file.read()),
                model="whisper-large-v3-turbo", # Либо "whisper-large-v3"
                response_format="json", # Можно использовать "text", "verbose_json", "srt", "vtt"
                language="en", # Задайте язык (например, "en" или "ru") или уберите для автоопределения
                temperature=0.0 # Опционально
            )
            
            print("\n--- Результат Транскрипции ---\n")
            print(transcription.text)
            return transcription.text
            
    except FileNotFoundError:
        print(f"Ошибка: Файл не найден по пути {filepath}. Пожалуйста, укажите верный путь.")
    except Exception as e:
        print(f"Произошла ошибка при транскрипции: {e}")

# Запуск функции для тестового аудио
if os.path.exists(audio_filepath):
    result_text = transcribe_audio(audio_filepath)
else:
    print(f"Пожалуйста, загрузите или разместите файл: {audio_filepath} для тестирования.")
```
